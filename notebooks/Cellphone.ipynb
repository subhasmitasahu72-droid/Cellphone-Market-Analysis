## Create count plots for all categorical variables to visualize their distributions.

```python
#import libraries
import math
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_style("whitegrid")#styling the plot
sns.set_palette("coolwarm")

categorical_cols = df.select_dtypes(include=['object']).columns.tolist() #automatically finding the categorical value

plt.figure(figsize=(25, len(categorical_cols) * 4)) #creating a plot 

for i, col in enumerate(categorical_cols, 1): #loop for each categorical column and enumerate() used for get index+value together
    #arrange plots in grid form
    cols = 3 #3 plots in a single row
    rows = math.ceil(len(categorical_cols) / cols) #for rows number needed and math.ceil function used for round up to the nearest integer
    plt.subplot(rows, cols, i) #final grid position 

    unique_count = df[col].nunique()     #count number of unique categories ,used for reduce the overcrowded countplot
#for many categories showing top 10
    if unique_count > 10:
        top_vals = df[col].value_counts().head(10)     #value_counts() use for counts of each category frequency
        sns.countplot(data=df
            [df[col].isin(top_vals.index)],        # removing all the rows from dataframe except top 10 variables
                      x=col,
                      order=top_vals.index)       #showing bars in a correct order
        plt.title(f"Top 10 {col} (Total {unique_count} categories)", fontsize=14, fontweight='bold')       #creating meaningful title
        
        # Add percentage labels
        counts = top_vals.values       # gives only numbers of count
        total = len(df)
        for j, value in enumerate(counts):  #only values index+count
            percent = round((value / total) * 100, 1)
            plt.text(j, value + 0.5, f"{percent}%", ha='center', fontsize=11)

    else:
        counts = df[col].value_counts()    #.value_count gives value with name
        sns.countplot(data=df, x=col, order=counts.index)
        plt.title(f"{col} (Total {unique_count})", fontsize=14, fontweight='bold')

        # Add percentage labels
        total = len(df)
        for j, (name, value) in enumerate(counts.items()):    #index+categorical name count
            percent = round((value / total) * 100, 1)
            plt.text(j, value + 0.5, f"{percent}%", ha='center', fontsize=11)

    plt.xticks(rotation=90, fontsize=11) #control the text on x axis 
    plt.xlabel(col, fontsize=12)
    plt.ylabel("Count", fontsize=12)

plt.suptitle("Categorical Variable Distributions", fontsize=20, fontweight='bold')
plt.tight_layout(rect=[0, 0, 1, 0.97]) #fixing overlaping text 
plt.show()
```

# 1. Hypothesis Testing

## 1a. Two-sample t-test: Salary difference between genders

```python
male_salary = df[df['gender']=='Male']['Salary_in_INR']
female_salary = df[df['gender']=='Female']['Salary_in_INR']
t_stat, p_value = stats.ttest_ind(male_salary, female_salary)
print("T-test: Salary difference between genders")
print("t-statistic:", t_stat, "p-value:", p_value, "\n")
```

```python
male_salary.mean(), female_salary.mean()
```

Interpretation---
-The average female salary (₹534k) is slightly higher than the average male salary (₹521k).
-The p-value = 0.0248 (< 0.05) indicates that this difference is statistically significant.
-There is a meaningful salary difference between genders — females earn more on average in this dataset.


## 1b. One-sample t-test: RAM > 16 GB

```python
t_stat, p_value = stats.ttest_1samp(df['RAM'], 16, alternative='greater')
print("One-sample t-test: Mean RAM > 16 GB")
print("t-statistic:", t_stat, "p-value:", p_value, "\n")
```

```python
df['RAM'].mean()
```

Interpretation---
-p-value = 1.0 (> 0.05), so we fail to reject the null hypothesis.
-There is no evidence that the mean RAM is greater than 16 GB.
-The negative t-statistic also shows that the actual mean RAM is much lower than 16 GB.

## 1c. One-sample t-test: Main camera > 12 MP


```python
t_stat, p_value = stats.ttest_1samp(df['main camera'], 12, alternative='greater')
print("One-sample t-test: Main camera > 12 MP")
print("t-statistic:", t_stat, "p-value:", p_value, "\n")
```

```python
df['main camera'].mean()
```

Interpretation---
-The p-value is extremely small (< 0.05), so we reject the null hypothesis.
-There is strong evidence that the average main camera resolution is greater than 12 MP.
-The very high positive t-statistic confirms that the mean camera MP is far above 12 MP.

```python
android_price.mean(), ios_price.mean()
```

Interpretation----

# 2. ANOVA Tests

## 2a. Price difference across brands

```python
brands_price = [df[df['brand']==b]['price(INR)'] for b in df['brand'].unique()]
f_stat, p_value = stats.f_oneway(*brands_price)
print("ANOVA: Price difference across brands")
print("F-statistic:", f_stat, "p-value:", p_value, "\n")
```

```python
df.groupby('brand')['price(INR)'].mean().sort_values()
```

Interpretation---
-There is a highly significant difference in average phone prices across different brands.
-Sony is the costliest brand, while Motorola is the cheapest. Premium brands (Oppo, Vivo, Apple, Samsung) have much higher average prices than mid-range/budget brands.

## 2b. Price difference across occupations

```python
occupations_price = [df[df['occupation']==o]['price(INR)'] for o in df['occupation'].unique()]
f_stat, p_value = stats.f_oneway(*occupations_price)
print("ANOVA: Price difference across occupations")
print("F-statistic:", f_stat, "p-value:", p_value, "\n")
```

Interpretation---
-The p-value (0.699) is much greater than 0.05, so we fail to reject the null hypothesis.
-This means smartphone prices do NOT significantly differ across occupations.
-Occupation does not have a meaningful impact on the price of the phone people buy in this dataset.

## 2c. Performance rating across age groups (<30, 30-50, >50)

```python
group1 = df[df['age']<30]['performance']
group2 = df[(df['age']>=30) & (df['age']<=50)]['performance']
group3 = df[df['age']>50]['performance']
f_stat, p_value = stats.f_oneway(group1, group2, group3)
print("ANOVA: Performance rating across age groups")
print("F-statistic:", f_stat, "p-value:", p_value, "\n")
```

Interpretation----
-The p-value (0.79) is much greater than 0.05, so we fail to reject the null hypothesis.
-This means performance ratings do NOT significantly differ across age groups.
-Age does not influence how users rate smartphone performance in this dataset.

## 2d. Battery size across brands

```python
brands_battery = [df[df['brand']==b]['battery size'] for b in df['brand'].unique()]
f_stat, p_value = stats.f_oneway(*brands_battery)
print("ANOVA: Battery size difference across brands")
print("F-statistic:", f_stat, "p-value:", p_value, "\n")
```

```python
df.groupby('brand')['battery size'].mean().sort_values()
```

Interpretation---
-There is a highly significant difference in battery sizes among different brands.
-The very large F-statistic indicates that brand strongly influences battery capacity.
-Some brands offer consistently larger batteries, while others offer smaller ones.

# 3. Chi-Square Tests

## 3a. OS choice vs gender

```python
os_gender_table = pd.crosstab(df['operating system'], df['gender'])
chi2, p, dof, expected = stats.chi2_contingency(os_gender_table)
print("Chi-square: OS choice vs Gender")
print("Chi2:", chi2, "p-value:", p, "\n")
```

Interpretation----
-p-value (0.977) is much greater than 0.05, so we fail to reject the null hypothesis.
-This means gender has no significant association with operating system choice.
-Male and female users choose Android and iOS in similar proportions.
-OS preference is independent of gender in this dataset.

## 3b. Brand choice vs OS

```python
brand_os_table = pd.crosstab(df['brand'], df['operating system'])
chi2, p, dof, expected = stats.chi2_contingency(brand_os_table)
print("Chi-square: Brand vs OS")
print("Chi2:", chi2, "p-value:", p, "\n")
```

Interpretation----
-p-value is extremely small, so we reject the null hypothesis.
-There is a very strong association between smartphone brand and operating system.
-The extremely large Chi-square value confirms that brand strongly determines the OS.

## 3c. Brand choice vs occupation

```python
brand_occ_table = pd.crosstab(df['brand'], df['occupation'])
chi2, p, dof, expected = stats.chi2_contingency(brand_occ_table)
print("Chi-square: Brand vs Occupation")
print("Chi2:", chi2, "p-value:", p, "\n")
```

Interpretation----
-The p-value (~1.0) is much greater than 0.05, so we fail to reject the null hypothesis.
-There is no significant association between occupation and the phone brand users buy.
-This means people from different jobs choose similar brands, and brand preference does not depend on occupation.
